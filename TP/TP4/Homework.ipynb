```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price < 0:
            print(new_price, "$", "The price can not be negative.")
        else:
            self.__price = new_price

    def show_info(self):
        print(self.title, "by", self.author, "- $", self.__price)

class Ebook(Book):
    def show_info(self):  
        print("Format: E-Book")
        Book.show_info(self)

book = Book("Clean Code", "Robert Martin", 35)
book.show_info()

book.set_price(30)
print("New price: $", book.get_price())

book.set_price(-19)
book.show_info()

eBook = Ebook("Clean Code", "Robert Martn", 20)
eBook.show_info()
```
