    a) Encapsulation
Encapsulation is the OOP priciple of building data that operate on it inside a class and hiding the data by making it private. If we want to change or use that data, we must create a diffrent method.

```python
class Account:
    def __init__(self):
        self.__balance = 200

    def get_balance(self):
        return self.__balance

acc = Account()
print(acc.get_balance())
    
```

    c) Inheritance
Inheritance is the OOP principle that has many related classes, we call child class and parent class. The clild class can use the method and attribute of parent class, but parent class can not.

```python
class Animal:                      # parent class
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")


class Dog(Animal):                 # child class
    def bark(self):
        print(f"{self.name} says woof!")


d = Dog("Rex")
d.eat()     
d.bark()    

a = Animal("Generic")
```

    d) Polymorphism
Polymorphism is the OOP principle that make the code short and easy to understand when that code has the same mothod name behave differently depending on the object.

```python
class Animal:
    def speak(self):
        print("Some sound")


class Dog(Animal):
    def speak(self):               
        print("Woof!")


class Cat(Animal):
    def speak(self):               
        print("Meow!")


animals = [Dog(), Cat(), Animal()]

for a in animals:
    a.speak()                      
```
